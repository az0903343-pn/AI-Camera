# AI Employee Monitoring - Kaggle Training

Is notebook se YOLOv8 fine-tune karein taake `person` aur `cell phone` ki detection aur behtar ho jaye (office CCTV angle ke liye).

**Steps:**
1. Kaggle par new notebook banayen aur yeh cells paste karein.
2. Kaggle Datasets me apna dataset add karein (Roboflow se export: YOLOv8 format, classes: `person`, `cell phone`).
3. Training ke baad `best.pt` download kar ke GitHub repo ke `models/` folder me daal dein.

Agar custom dataset nahi hai to yeh notebook optional hai - app default `yolov8n.pt` par bhi kaam karega (COCO me person=0, cell phone=67 already hain).

```python
# Cell 1: install
!pip install ultralytics -q
```

```python
# Cell 2: dataset setup
# Option A: Kaggle dataset already attached hai (right side panel -> Add Input)
#   e.g. /kaggle/input/my-office-dataset/data.yaml
# Option B: Roboflow se download:
#
# !pip install roboflow -q
# from roboflow import Roboflow
# rf = Roboflow(api_key="YOUR_API_KEY")
# project = rf.workspace("your-workspace").project("employee-monitoring")
# dataset = project.version(1).download("yolov8")

DATA_YAML = "/kaggle/input/my-office-dataset/data.yaml"  # <-- apne dataset ka path
```

```python
# Cell 3: train (GPU T4/P100 select karein: Settings -> Accelerator)
from ultralytics import YOLO

model = YOLO("yolov8n.pt")  # n = fast, s/m/l = zyada accurate

results = model.train(
    data=DATA_YAML,
    epochs=50,
    imgsz=640,
    batch=16,
    patience=10,
    device=0,
)
```

```python
# Cell 4: validate + best model location
metrics = model.val()
print("mAP50-95:", metrics.box.map)

# best.pt ka path usually:
# /kaggle/working/runs/detect/train/weights/best.pt
!ls /kaggle/working/runs/detect/train/weights/
```
